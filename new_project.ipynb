#COIN TOSS SIMULTATOR

```python
import numpy as np

# Number of coin tosses
num_flips = 500

# Simulate coin flips: 'H' for heads, 'T' for tails
flips = np.random.choice(['H', 'T'], size=num_flips)

# Count heads and tails
num_heads = np.sum(flips == 'H')
num_tails = np.sum(flips == 'T')

# Calculate percentages
perc_heads = (num_heads / num_flips) * 100
perc_tails = (num_tails / num_flips) * 100

# Print results
print(f"Total tosses: {num_flips}")
print(f"Heads: {num_heads} ({perc_heads:.2f}%)")
print(f"Tails: {num_tails} ({perc_tails:.2f}%)")
```
